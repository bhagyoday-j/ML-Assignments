```python
def say_hi() :
  return "hello there"

def say_greet() :
  return "good evening"

def uppercase_decorator(function) :
  def wrapper() :
    func = function()
    make_uppercase = func.upper()
    return make_uppercase

  return wrapper


decorator = uppercase_decorator(say_hi)
decorator()
decorator1 = uppercase_decorator(say_greet)
decorator1()
```

```python
from itertools import zip_longest
name = ["dada", "mama", "baba"]
info = ["9850", "6032", "3297"]

for nm, info in zip_longest(name, info) :
  print(nm, info)
```
