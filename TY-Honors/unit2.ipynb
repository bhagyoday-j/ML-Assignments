```python
def say_hi() :
  return "hello there"

def say_greet() :
  return "good evening"

def uppercase_decorator(function) :
  def wrapper() :
    func = function()
    make_uppercase = func.upper()
    return make_uppercase

  return wrapper


decorator = uppercase_decorator(say_hi)
decorator()
decorator1 = uppercase_decorator(say_greet)
decorator1()
```

```python
# ==========================================================
# Pandas Assignment - Drop Columns, Drop Rows and iloc
# ==========================================================

import pandas as pd
import numpy as np

# ----------------------------------------------------------
# Create DataFrame
# ----------------------------------------------------------

technologies = {
    "COURSES": ["SPARRK", "PYSPARK", "HADOOOP", "PYTHON",
                "PANDAS", None, "SPARK", "PYTHON"],

    "FEES": [22000, 25000, 23000, 24000,
             np.nan, 25000, 25000, 22000],

    "DURATION": ["30", "50", "55", "40",
                 "60", "35", "", "50"],

    "DISCOUNT": [1000, 2300, 1000, 1200,
                 2500, 1300, 1400, 1600]
}

row_labels = ['r0', 'r1', 'r2', 'r3', 'r4', 'r5', 'r6', 'r7']

df = pd.DataFrame(technologies, index=row_labels)

print("\nORIGINAL DATAFRAME")
print(df)

# ==========================================================
# DROPPING SINGLE COLUMN
# ==========================================================

print("\n1. Drop column FEES")
df2 = df.drop(["FEES"], axis=1)
print(df2)

print("\n2. Drop column FEES using labels")
df2 = df.drop(labels=["FEES"], axis=1)
print(df2)

print("\n3. Drop column FEES using columns parameter")
df2 = df.drop(columns=["FEES"])
print(df2)

print("\n4. Drop column using column index 1")
df2 = df.drop(df.columns[1], axis=1)
print(df2)

print("\n5. One-line drop")
print(df.drop(df.columns[1], axis=1))

# ==========================================================
# DROP COLUMN FROM ORIGINAL DATAFRAME USING inplace=True
# ==========================================================

print("\n6. Drop DURATION column permanently")

df_copy = pd.DataFrame(technologies)

df_copy.drop(df_copy.columns[2], axis=1, inplace=True)

print(df_copy)

# ==========================================================
# DELETE MULTIPLE COLUMNS
# ==========================================================

print("\n7. Delete COURSES and FEES columns")

df2 = df.drop(["COURSES", "FEES"], axis=1)

print(df2)

print("\n8. Delete first two columns using indexes")

df2 = df.drop(df.columns[[0, 1]], axis=1)

print(df2)

print("\n9. Delete columns using list")

liscol = ["COURSES", "FEES"]

df2 = df.drop(liscol, axis=1)

print(df2)

# ==========================================================
# DELETE MULTIPLE ROWS
# ==========================================================

print("\n10. Delete rows r0 and r3")

df1 = df.drop(['r0', 'r3'], axis=0)

print(df1)

# ==========================================================
# ILOC OPERATIONS
# ==========================================================

print("\n11. All rows but first 2 columns only")
df2 = df.iloc[:, 0:2]
print(df2)

print("\n12. First 2 rows and all columns")
df2 = df.iloc[0:2, :]
print(df2)

print("\n13. First 3 rows and first 3 columns")
df2 = df.iloc[0:3, 0:3]
print(df2)

print("\n14. Excluding first 3 rows and first 3 columns")
df2 = df.iloc[3:, 3:]
print(df2)

print("\n15. Row number 1 and column 1 & 2 only")
df3 = df.iloc[1:2, 1:3]
print(df3)

print("\n16. Select row number 2")
df2 = df.iloc[2]
print(df2)

print("\n17. Select rows by index list [2,3,6]")
df2 = df.iloc[[2, 3, 6]]
print(df2)

print("\n18. Select rows from index 1 to 4")
df2 = df.iloc[1:5]
print(df2)

print("\n19. First row")
df2 = df.iloc[:1]
print(df2)

print("\n20. First 3 rows")
df2 = df.iloc[:3]
print(df2)

print("\n21. Last row")
df2 = df.iloc[-1:]
print(df2)

# ==========================================================
# END OF PROGRAM
# ==========================================================
```
