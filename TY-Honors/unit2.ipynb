```python
def add(a, b, c) :
  return a+b+c
print(add(1, 2, 3))

add1 = lambda a, b, c : a + b + c
add1(4, 5, 6)
```

```python
def mul(a, b, c) :
  multi = a*b*c
  return multi

print(mul(2, 3, 1))

mul1 = lambda a, b, c : a * b * c
mul(2, 3, 4)
```

```python
val = lambda *args : sum(args)
print(val(1, 2, 3, 4, 5, 6))
val(1, 2, 3, 4, 5, 6, 7, 8, 9)
```

```python
def myfun(*args) :
  for i in args :
    print(i)

print("Python", "CPP", "JS", "Java")
```

```python
def person(name, *data) :
  print(name)
  print(data)

person('Navin', 28, 'Mumbai', 323)
```
