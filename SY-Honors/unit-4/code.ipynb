```python
import numpy as np 

v2 = np.array([2, 3])
v3 = np.array([1, 2, 3])
vn = np.array([1, 2, 3, 4, 5])

print("2D", v2)
print("3D", v3)
print("nD", vn)
```

```python
row = np.array([[1, 2, 3]])
col = np.array([[1], [2], [3]])
print("Row vector:", row)
print("Column vector:", col)
```

```python
# Dot product of two vectors => sum of the products of corresponding elements
a = np.array([1, 2])
b = np.array([3, 4])

dot = np.dot(a, b)
angle = np.arccos(dot / (np.linalg.norm(a) * np.linalg.norm(b)))#linalg.norm() calculates the magnitude of a vector

print("Dot product:", dot)
print("Angle (radians):", angle)
```

```python
#unit vector and projection
#Unit vector => vector with magnitude of 1
#Projection of a onto b => (dot product of a and b) / (magnitude of b) * (unit vector of b) => one vector on another vector

b = np.array([1, 2])
a = np.array([3, 4])

unit = a / np.linalg.norm(a)
projection1 = (np.dot(a, b) / np.linalg.norm(b)) * (b / np.linalg.norm(b))
projection2 = (np.dot(a, b) / np.dot(b, b)) * b

print("Unit vector of a:", unit)
print("Projection of a onto b:", projection1)
print("Projection of a onto b (alternative):", projection2)
```

```python
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 50)
y = 2*x + 1

plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("y")

plt.title("Line: y = 2x + 1")

plt.show()
```

```python
#plane ax + by + cz = d

x,  y, z = 1, 1, 1
value = 2*x + 3*y + 4*z + 5
print("Value of ax + by + cz:", value)
```

```python
x0, y0, z0 = 1, 2, 3

distance = abs(2*x0 + 3*y0 + 4*z0 + 5) / np.sqrt(2**2 + 3**2 + 4**2)

print("Distance from point to plane:", distance)
```

```python
# Matrices

I = np.eye(3)
D = np.diag([1, 2, 3])
S = 5*np.eye(3)

print("Identity matrix:\n", I)
print("Diagonal matrix:\n", D)
print("Scaled identity matrix:\n", S)
```

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print("Matrix A:\n", A)
print("Matrix B:\n", B)

print("Multiplication of A and B:\n", np.dot(A, B))
print("Transpose of A:\n", np.transpose(A))
print("Transpose of A:\n", A.T)
```

```python
# Derivatives
from sympy.misc import derivative

f = lambda x: x**2 

print("Derivative of f at x=2:", derivative(f, 2.0))
```

```python
from scipy.optimization import minimize

f = lambda x: x**2 

res = minimize(f, 0.0)  

print("Minimum value of f:", res.fun)
print("Value of x that minimizes f:", res.x)
```
