```python
#np.linspace(start, stop, num) => start is the starting value of the sequence, stop is the end value of the sequence, and num is the number of samples to generate between start and stop (inclusive).

from scipy import stats
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-3, 3, 100)
pdf = stats.norm.pdf(x)
plt.plot(x, pdf)
plt.title("Normal Distribution PDF")
plt.show()
```

```python
from scipy.optimize import minimize

def f(x):
    return (x - 3) ** 2

result = minimize(f, 0)
print("Minimum near:", result.x)
```
