## 1️⃣2️⃣ List vs NumPy Performance

```python
#List uses explicit loop. NumPy uses C-based vectorization. NumPy is significantly faster.Shows importance of mathematical engine

import time

# Python List computation
lst = list(range(1000000))
start = time.time()
lst = [x*2 for x in lst]
print("List Time:", time.time() - start)

# NumPy computation
arr_np = np.arange(1000000)
start = time.time()
arr_np = arr_np * 2
print("NumPy Time:", time.time() - start)
```

# 2. Generate and visualize normal distributions using SciPy.

```python
# Import required libraries
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

#Generate Normal Distribution Data
# Set mean and standard deviation
mean = 70        # average marks
std_dev = 10     # spread of marks

# Generate 1000 random samples from normal distribution
data = np.random.normal(mean, std_dev, 1000)

# Print first 10 values
print("Sample generated data:", data[:10])
```

```python
#Visualize Histogram of Data
# Plot histogram
plt.hist(data, bins=30, density=True)

# Labels
plt.title("Histogram of Normally Distributed Data")
plt.xlabel("Marks")
plt.ylabel("Probability Density")

plt.show()
#YOU will see a bell-shaped curve pattern.
```

```python
# Plot Probability Density Function (PDF)
# Generate x values
x = np.linspace(30, 110, 100)

# Calculate normal PDF
pdf = norm.pdf(x, mean, std_dev)

# Plot PDF curve
plt.plot(x, pdf)

plt.title("Normal Distribution PDF")
plt.xlabel("Marks")
plt.ylabel("Probability Density")

plt.show()
```

```python
#Combine Histogram + PDF
# Plot histogram of generated data
plt.hist(data, bins=30, density=True)

# Generate x values
x = np.linspace(min(data), max(data), 100)

# Plot theoretical PDF
plt.plot(x, norm.pdf(x, mean, std_dev))

# Labels
plt.title("Normal Distribution Visualization")
plt.xlabel("Marks")
plt.ylabel("Density")

plt.show()
# This visualization shows: Histogram → real generated data . Curve → theoretical normal distribution
```

```python
#Calculate Statistics
# Calculate statistics
print("Mean:", np.mean(data))
print("Standard Deviation:", np.std(data))
```

## Lecture 26 – Random Experiments
Data scientists use it for:

Example:

If 10 students flip a coin:

| Flip | Result |
| ---- | ------ |
| 1    | H      |
| 2    | T      |
| 3    | H      |
| 4    | H      |

Heads may not be exactly 50%.

But if we flip 1000 times, the result will be very close to 0.5.

This idea is called the Law of Large Numbers, which is very important in Data Science.


```python

import numpy as np

trials = 1000
coin = np.random.choice(["H","T"], size=trials)

prob_heads = np.sum(coin == "H") / trials
print("Estimated Probability of Heads:", prob_heads)
```

# Lecture 35: Binomial Distribution
Binomial Distribution describes the probability of getting a certain number of successes in a fixed number of independent trials.

Each trial has only two possible outcomes:

Success (1)

Failure (0)
Examples:
| Experiment           | Success | Failure  |
| -------------------- | ------- | -------- |
| Coin toss            | Head    | Tail     |
| Email classification | Spam    | Not spam |
| Exam result          | Pass    | Fail     |
A problem follows binomial distribution if:

1️⃣ Fixed number of trials (n)
2️⃣ Only two outcomes (success or failure)
3️⃣ Probability of success (p) remains constant
4️⃣ Trials are independent

Example:

Toss a coin 5 times and count how many heads occur.
Binomial distribution is used in many Data Science tasks:
| Application          | Example              |
| -------------------- | -------------------- |
| Email spam detection | Spam vs Not spam     |

| A/B testing          | Success vs Failure   |
Binomial Distribution helps us model binary events repeated multiple times.

It is widely used in:

Machine Learning

Classification models

A/B testing

Prediction systems

## Lecture 40 – Exponential Distribution

## Lecture 41 – Uniform Distribution
A Uniform Distribution means all outcomes have equal probability.

In simple words:

👉 Every value in the range is equally likely to occur.

Uniform distribution represents complete randomness where all outcomes are equally likely, and it is widely used in simulations, machine learning, and random sampling.

```python
from scipy.stats import uniform

a = 0
b = 10

x = np.linspace(a, b, 200)
pdf = uniform.pdf(x, loc=a, scale=b-a)

plt.figure()
plt.plot(x, pdf)
plt.title("Uniform Distribution PDF")
plt.xlabel("x")
plt.ylabel("Density")
plt.show()
```

# Lecture 51: Level of Significance and P-Value
Step 1: Define Hypothesis
H0: μ = 100
H1: μ ≠ 100

Given:

Sample Mean (x̄) = 105

Population Mean (μ) = 100

Population Std Dev (σ) = 15

Sample Size (n) = 36

Level of Significance (α) = 0.05

Interpretation
If p-value < α (0.05), reject H0.
Otherwise, fail to reject H0.

This demonstrates how the level of significance controls the probability of Type I Error.

```python

# Given values
x_bar = 105
mu = 100
sigma = 15
n = 36
alpha = 0.05

# Step 2: Compute Z-statistic
z = (x_bar - mu) / (sigma / np.sqrt(n))
print("Z-statistic:", z)

# Step 3: Compute two-tailed P-value
p_value = 2 * (1 - norm.cdf(abs(z)))
print("P-value:", p_value)

# Step 4: Decision Rule
if p_value < alpha:
    print("Reject the Null Hypothesis (H0)")
else:
    print("Fail to Reject the Null Hypothesis (H0)")

```
